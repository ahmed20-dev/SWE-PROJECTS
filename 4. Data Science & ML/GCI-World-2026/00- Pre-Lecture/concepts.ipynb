## 1. Python basics

````python
name = "Ada"          # str
age = 30              # int
height = 1.68         # float
is_student = True     # bool
nothing = None        # NoneType

print(type(age), type(height), type(is_student), type(nothing))
print(f"{name} is {age} years old and {height:.1f} m tall")   # f-string with formatting

# multiple assignment and swapping
a, b = 1, 2
a, b = b, a
print(a, b)

# type conversion
print(int("42") + 1, float("3.5"), str(99) + "!", bool(0), bool("text"))
````

### Conditionals and truthiness
Falsy values: `0`, `0.0`, `""`, `[]`, `{}`, `()`, `set()`, `None`, `False`. Everything else is truthy.

````python
score = 82
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
else:
    grade = "C"
print(grade)

# comparison chaining, logic operators, ternary
print(0 < score < 100, score > 90 or score < 85, not score == 82)
label = "pass" if score >= 50 else "fail"
print(label)

items = []
if not items:
    print("empty list is falsy")
````

## 2. Core data structures
### List: ordered, mutable

````python
nums = [3, 1, 4, 1, 5]
nums.append(9)
nums.insert(0, 100)
nums.extend([7, 8])
print(nums)
print(nums[0], nums[-1], nums[1:4], nums[::-1], nums[::2])
nums.remove(100)          # removes first matching value
last = nums.pop()         # removes and returns last
print(last, nums)
print(sorted(nums), nums)          # sorted() returns a NEW list
nums.sort(reverse=True)            # sort() changes in place
print(nums, len(nums), sum(nums), nums.count(1), nums.index(4))

# copying trap
a = [1, 2, 3]
b = a            # same object!
c = a.copy()     # independent copy
b.append(99)
print(a, c)
````

### Tuple: ordered, immutable

````python
point = (3, 4)
x, y = point                     # unpacking
print(x, y)
first, *rest = [10, 20, 30, 40]
print(first, rest)
# point[0] = 9   # TypeError: tuples cannot be modified
def min_max(v): return min(v), max(v)   # returning a tuple
lo, hi = min_max([4, 2, 9])
print(lo, hi)
````

### Dictionary: key to value

````python
person = {"name": "Ada", "age": 30}
person["city"] = "London"
print(person["name"], person.get("email", "N/A"))
print(person.keys(), person.values())
for key, value in person.items():
    print(key, "->", value)
person.update({"age": 31, "job": "Engineer"})
removed = person.pop("job")
print("city" in person, removed, len(person))

# counting pattern
counts = {}
for ch in "mississippi":
    counts[ch] = counts.get(ch, 0) + 1
print(counts)

# nested
data = {"ada": {"scores": [90, 85]}, "bob": {"scores": [70, 75]}}
print(data["ada"]["scores"][1])
````

### Set: unique, unordered

````python
tags = {"ml", "ai", "ml"}
print(tags)
print(set([1, 2, 2, 3, 3]))               # dedupe a list
A, B = {1, 2, 3}, {2, 3, 4}
print(A & B, A | B, A - B, A ^ B)          # intersection, union, difference, symmetric diff
print(2 in A, A.issubset({1, 2, 3, 4}))
tags.add("nlp"); tags.discard("ai")
print(tags)
````

## 3. Comprehensions, functions, lambdas
### Comprehensions

````python
squares = [x**2 for x in range(10)]
evens   = [x for x in range(10) if x % 2 == 0]
labels  = ["even" if x % 2 == 0 else "odd" for x in range(4)]
pairs   = [(i, j) for i in range(2) for j in range(2)]
word_len = {w: len(w) for w in ["data", "science"]}
unique_lens = {len(w) for w in ["aa", "bb", "ccc"]}
total = sum(x for x in range(5))      # generator expression (lazy, no list built)
print(squares, evens, labels, pairs, word_len, unique_lens, total, sep="\n")
````

### Functions

````python
def normalize(values, method="minmax"):
    """Scale a list of numbers to 0..1. (This is a docstring.)"""
    if method == "minmax":
        lo, hi = min(values), max(values)
        return [(v - lo) / (hi - lo) for v in values]
    raise ValueError(f"Unknown method: {method}")

print(normalize([1, 2, 3, 4, 5]))
print(normalize(values=[10, 20, 30]))     # keyword argument

def show(*args, **kwargs):               # variable arguments
    print("args  :", args)
    print("kwargs:", kwargs)
show(1, 2, a=3, b=4)

nums = [1, 2, 3]
show(*nums, **{"k": "v"})                 # unpacking into a call

# mutable default argument trap
def bad(item, bucket=[]):
    bucket.append(item); return bucket
print(bad(1), bad(2))                     # [1] then [1, 2]: shared list!
def good(item, bucket=None):
    bucket = [] if bucket is None else bucket
    bucket.append(item); return bucket
print(good(1), good(2))

# scope
counter = 0
def inc():
    global counter
    counter += 1
inc(); inc()
print(counter)

# type hints (documentation, not enforced)
def area(w: float, h: float) -> float:
    return w * h
print(area(2, 3.5))
````

### Lambda, map, filter, sorted with key

````python
double = lambda x: x * 2
print(double(4))
print(list(map(lambda x: x * 2, [1, 2, 3])))
print(list(filter(lambda x: x > 1, [1, 2, 3])))

people = [("Ada", 30), ("Bob", 25), ("Cy", 41)]
print(sorted(people, key=lambda p: p[1]))
print(sorted(people, key=lambda p: p[1], reverse=True)[0])
print(max(people, key=lambda p: p[1]))

words = ["banana", "fig", "apple"]
print(sorted(words, key=len), sorted(words, key=str.lower))
````

## 4. Classes, errors, files
### Classes
A class bundles **data** (attributes) and **behavior** (methods). scikit-learn's `fit` / `transform` API uses exactly this pattern.

````python
class Scaler:
    def __init__(self):
        self.mean_ = None

    def fit(self, data):
        self.mean_ = sum(data) / len(data)
        return self                      # returning self allows chaining

    def transform(self, data):
        return [d - self.mean_ for d in data]

s = Scaler().fit([1, 2, 3])
print(s.mean_, s.transform([1, 2, 3]))

class Account:
    bank = "PyBank"                      # class attribute shared by all instances

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def __repr__(self):                  # how the object prints
        return f"Account({self.owner!r}, {self.balance})"

class Savings(Account):                  # inheritance
    def add_interest(self, rate):
        self.balance *= 1 + rate

acc = Savings("Ada", 100)
acc.deposit(50)
acc.add_interest(0.10)
print(acc, acc.bank, isinstance(acc, Account))
````

### Errors and exceptions

````python
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print("Error:", e)
else:
    print("no error, result =", result)
finally:
    print("always runs")

def to_int(text):
    try:
        return int(text)
    except (ValueError, TypeError):
        return None

print(to_int("12"), to_int("abc"), to_int(None))

# raising your own errors
def withdraw(balance, amount):
    if amount > balance:
        raise ValueError("insufficient funds")
    return balance - amount

try:
    withdraw(10, 50)
except ValueError as e:
    print("Caught:", e)

# custom exception
class ValidationError(Exception):
    pass

try:
    raise ValidationError("age must be positive")
except ValidationError as e:
    print(type(e).__name__, e)

# common errors: KeyError, IndexError, TypeError, FileNotFoundError
for bad in (lambda: {}["x"], lambda: [][0], lambda: "a" + 1, lambda: open("missing.txt")):
    try: bad()
    except Exception as e: print(type(e).__name__, "->", e)
````

### Files
Always use `with`: it closes the file even if an error occurs.
Modes: `r` read, `w` write (overwrites), `a` append.

````python
with open("notes.txt", "w") as f:
    f.write("hello\n")
    f.write("second line\n")
    f.writelines(["third\n", "fourth\n"])

with open("notes.txt", "a") as f:
    f.write("appended\n")

with open("notes.txt") as f:
    content = f.read()
print(content)

with open("notes.txt") as f:
    for line in f:                       # memory-friendly line-by-line
        print(repr(line.strip()))

with open("notes.txt") as f:
    lines = f.read().splitlines()
print(lines)

# safer paths with pathlib
from pathlib import Path
p = Path("notes.txt")
print(p.exists(), p.suffix, p.stem, p.stat().st_size, "bytes")
print(p.read_text().count("\n"), "lines")

# JSON is the common way to store dicts
import json
with open("cfg.json", "w") as f:
    json.dump({"lr": 0.01, "layers": [64, 32]}, f, indent=2)
with open("cfg.json") as f:
    print(json.load(f))

# modules
import math
from collections import Counter, defaultdict
print(math.sqrt(16), Counter("mississippi").most_common(2))
dd = defaultdict(list); dd["a"].append(1); print(dict(dd))

import os
os.remove("notes.txt"); os.remove("cfg.json")
````

## 5. datetime
Main classes: `date` (no time), `time`, `datetime` (both), `timedelta` (a duration).

````python
from datetime import date, time, datetime, timedelta

d = date(2026, 3, 15)
print(d, d.year, d.month, d.day, d.weekday(), d.isoweekday())   # Monday=0 ... Sunday=6
dt = datetime(2026, 3, 15, 14, 30, 45)
print(dt, dt.hour, dt.minute, dt.date(), dt.time())
print(date.today(), datetime.now())
````

### Parsing and formatting: `strptime` (string to datetime) and `strftime` (datetime to string)

| Code | Meaning | Code | Meaning |
|---|---|---|---|
| `%Y` | 4-digit year | `%H` | hour 00-23 |
| `%m` | month 01-12 | `%M` | minute |
| `%d` | day 01-31 | `%S` | second |
| `%B` / `%b` | month name / short | `%A` / `%a` | weekday name / short |
| `%j` | day of year | `%p` | AM/PM |

````python
parsed = datetime.strptime("15/03/2026 14:30", "%d/%m/%Y %H:%M")
print(parsed)
print(parsed.strftime("%A, %d %B %Y"))
print(parsed.strftime("%I:%M %p"))
print(parsed.isoformat())
print(date.fromisoformat("2026-03-15"), datetime.fromisoformat("2026-03-15T14:30:00"))
````

### Arithmetic with `timedelta`

````python
d1, d2 = date(2026, 1, 1), date(2026, 3, 15)
gap = d2 - d1                            # a timedelta
print(gap, gap.days)
print(d1 + timedelta(days=45))
print(datetime(2026, 3, 15, 23, 0) + timedelta(hours=3, minutes=30))
print(timedelta(days=1, hours=12).total_seconds())
print(d1 < d2, max(d1, d2))

# useful recipes
def add_months(d, n):
    y, m = divmod(d.month - 1 + n, 12)
    import calendar
    last = calendar.monthrange(d.year + y, m + 1)[1]
    return date(d.year + y, m + 1, min(d.day, last))
print(add_months(date(2026, 1, 31), 1))       # clamps to end of February

first_of_month = date(2026, 3, 15).replace(day=1)
print(first_of_month)
````

### Time zones
Naive datetimes have no zone; aware ones do. Prefer aware datetimes (or UTC) for anything shared across regions.

````python
from datetime import timezone
from zoneinfo import ZoneInfo

utc_now = datetime(2026, 3, 15, 12, 0, tzinfo=timezone.utc)
print(utc_now)
try:
    print(utc_now.astimezone(ZoneInfo("Africa/Mogadishu")))
    print(utc_now.astimezone(ZoneInfo("America/New_York")))
except Exception as e:
    print("zoneinfo data missing:", e, "(pip install tzdata)")
````

## 6. csv
The `csv` module reads and writes comma-separated files. **Everything read from a CSV is a string**, so convert numbers and dates yourself.
Always open with `newline=""` and `encoding="utf-8"`.

````python
import csv

rows = [
    ["name", "dept", "salary", "hired"],
    ["Ada", "Eng", 120, "2024-01-15"],
    ["Bob", "Eng", 95, "2025-06-01"],
    ["Cy", "Ops", 70, "2023-11-20"],
]
with open("staff.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(rows)

print(open("staff.csv").read())
````

### Reading with `csv.reader` (lists) and `csv.DictReader` (dicts, preferred)

````python
with open("staff.csv", newline="", encoding="utf-8") as f:
    reader = csv.reader(f)
    header = next(reader)
    print(header)
    for row in reader:
        print(row, type(row[2]))          # salary is a str!

with open("staff.csv", newline="", encoding="utf-8") as f:
    people = list(csv.DictReader(f))
print(people[0])
print(people[0]["name"], int(people[0]["salary"]) + 1)
````

### Cleaning types and aggregating

````python
from datetime import datetime
from collections import defaultdict

for p in people:
    p["salary"] = int(p["salary"])
    p["hired"] = datetime.strptime(p["hired"], "%Y-%m-%d").date()

by_dept = defaultdict(list)
for p in people:
    by_dept[p["dept"]].append(p["salary"])
print({d: sum(v) / len(v) for d, v in by_dept.items()})

today = date(2026, 3, 15)
for p in people:
    print(p["name"], "tenure (days):", (today - p["hired"]).days)
````

### Writing with `csv.DictWriter`

````python
out = [{"name": p["name"], "tenure_years": round((today - p["hired"]).days / 365.25, 1)} for p in people]
with open("tenure.csv", "w", newline="", encoding="utf-8") as f:
    w = csv.DictWriter(f, fieldnames=["name", "tenure_years"])
    w.writeheader()
    w.writerows(out)
print(open("tenure.csv").read())

# other delimiters and quoting
with open("semi.csv", "w", newline="") as f:
    csv.writer(f, delimiter=";").writerow(["a,b", 'say "hi"', 3])
print(open("semi.csv").read())          # the module handles quoting for you

import os
for fn in ("staff.csv", "tenure.csv", "semi.csv"): os.remove(fn)
````

### When to leave `csv` for pandas
`csv` is perfect for small, simple, dependency-free jobs. For analysis use `pd.read_csv("file.csv", parse_dates=["hired"])`, which handles types, missing values, and dates for you.

## Cheat sheet
| Topic | Remember |
|---|---|
| Mutable vs immutable | list, dict, set change in place; str, tuple, int don't |
| Copy | `b = a` shares; `a.copy()` duplicates |
| Defaults | never use `[]` or `{}` as a default argument |
| Files | always `with open(...)`; CSV values are strings |
| Dates | `strptime` parses, `strftime` formats, `timedelta` does math |
| Errors | catch specific exceptions, raise clear ones |
